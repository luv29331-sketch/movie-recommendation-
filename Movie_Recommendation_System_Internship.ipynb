# Movie Recommendation System
### Content-Based Recommendation using TF-IDF and Cosine Similarity

This notebook organizes the supplied Movie Recommendation System into a cleaner, internship-ready workflow. It uses the same methodology: selected movie metadata → combined text features → TF-IDF → cosine similarity → title matching → recommendations.

**Dataset required:** `movies.csv` (the same dataset expected by the supplied notebook).

## 1. Import Libraries
The project uses Pandas and NumPy for data handling, scikit-learn for TF-IDF and cosine similarity, and `difflib` for title matching.

```python
import numpy as np
import pandas as pd
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## 2. Load and Inspect the Dataset

```python
# Place the same movies.csv used by the supplied project in this folder.
movies = pd.read_csv('movies.csv')
print('Dataset shape:', movies.shape)
display(movies.head())
```

The supplied project reports **4,803 rows and 24 columns**. The recommendation calculation uses five metadata fields: `genres`, `keywords`, `tagline`, `cast`, and `director`.

```python
selected_features = ['genres', 'keywords', 'tagline', 'cast', 'director']
print('Selected features:', selected_features)
```

## 3. Handle Missing Values

```python
for feature in selected_features:
    movies[feature] = movies[feature].fillna('')

print(movies[selected_features].isnull().sum())
```

## 4. Create the Combined Movie Feature
The five selected metadata fields are concatenated into one text representation for each movie.

```python
combined_features = (
    movies['genres'] + ' ' +
    movies['keywords'] + ' ' +
    movies['tagline'] + ' ' +
    movies['cast'] + ' ' +
    movies['director']
)

print('Number of combined feature records:', len(combined_features))
display(combined_features.head())
```

## 5. TF-IDF Vectorization

```python
vectorizer = TfidfVectorizer()
feature_vectors = vectorizer.fit_transform(combined_features)
print('Feature-vector matrix shape:', feature_vectors.shape)
```

## 6. Cosine Similarity

```python
similarity = cosine_similarity(feature_vectors)
print('Similarity matrix shape:', similarity.shape)
print('First diagonal value:', similarity[0, 0])
```

For the supplied dataset, the project documentation reports a **4,803 × 4,803** similarity matrix. Diagonal values are 1.0 because each movie is identical to itself in the feature space.

## 7. Movie Title Matching

```python
list_of_all_titles = movies['title'].tolist()

movie_name = input('Enter your favourite movie name: ')
find_close_match = difflib.get_close_matches(movie_name, list_of_all_titles)

if not find_close_match:
    raise ValueError('No close movie title was found in the dataset.')

close_match = find_close_match[0]
print('Matched movie:', close_match)
```

## 8. Generate Recommendations

```python
movie_index = movies[movies.title == close_match].index[0]
similarity_score = list(enumerate(similarity[movie_index]))
sorted_similar_movies = sorted(similarity_score, key=lambda x: x[1], reverse=True)

print(f'\nRecommendations similar to: {close_match}\n')
for i, (index, score) in enumerate(sorted_similar_movies[1:11], start=1):
    print(f'{i}. {movies.iloc[index].title}  | similarity={score:.4f}')
```

## 9. Method Summary
1. Load `movies.csv`.
2. Select `genres`, `keywords`, `tagline`, `cast`, and `director`.
3. Replace missing selected values with empty strings.
4. Combine the five fields into `combined_features`.
5. Convert the text into TF-IDF vectors.
6. Compute cosine similarity between movie vectors.
7. Match the user's title using `difflib`.
8. Rank movies by similarity and display recommendations.

**Important:** This is a content-based academic prototype. It does not use collaborative filtering or user-rating history.
